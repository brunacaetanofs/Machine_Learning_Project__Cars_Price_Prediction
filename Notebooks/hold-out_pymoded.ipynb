## Abstract


This project addresses the problem of automated vehicle price estimation through the development of a supervised machine learning system capable of predicting market prices based on user-provided attributes. The primary objective was to maximize the coefficient of determination ($R^2$) while minimizing the mean absolute error (MAE), ensuring both high explanatory power and low prediction error.

A complete Data Science workflow was implemented, including data cleaning, feature engineering, and preprocessing. Domain-driven features such as vehicle age, mileage intensity, and power efficiency were introduced to capture non-linear depreciation patterns. Numerical variables were processed using RobustScaler and KNNImputer, while categorical variables were encoded using Target Encoding.

Multiple regression models were evaluated and optimized using RandomizedSearchCV. The best-performing models, Gradient Boosting, Random Forest, Extra Trees, HistGradientBoosting, and Multilayer Perceptron, were combined into a Stacking Regressor with a RidgeCV meta-learner. The ensemble approach demonstrated superior performance compared to individual models, achieving a lower MAE on unseen data.

Feature importance analysis indicated that vehicle model, age-related variables, and power efficiency are the most influential factors in determining vehicle price.


## I. Identifying Business Needs

**Overview and Main Goals of the Project**


The "Cars 4 You" company currently faces a significant operational bottleneck in its vehicle valuation process. Market price estimation relies heavily on physical mechanical inspections, which has resulted in long waiting lists and delays in responding to potential customers. As a consequence, a relevant portion of interested clients turns to competing platforms that offer faster price estimates.

This dependency on in-person inspections limits the scalability of the business and increases operational costs, while also reducing customer satisfaction due to long response times. In a highly competitive market, the inability to provide quick and reliable price estimates represents a critical disadvantage.

To address this challenge, there is a clear business need for an automated solution capable of providing accurate vehicle price estimations based solely on information supplied by the customer. Such a solution would allow Cars 4 You to significantly reduce waiting times, improve customer experience, and optimize internal resources, while maintaining reliable and consistent pricing decisions.

From a business perspective, the success of this solution depends on its ability to deliver precise price estimates, minimizing large valuation errors that could negatively impact trust and decision-making. Therefore, accuracy and robustness are key requirements for the proposed system.


**Description of the Overall Process**

We implemented a comprehensive Data Science workflow consisting of the following sequential stages:

1. Data Cleaning & Preprocessing (Initial Phase):
    * Handling Strange Values: We began by identifying and treating impossible or "strange" values to ensure data quality.
    * Categorical Cleaning: We applied fuzzy matching techniques to standardize categorical entries and reduce cardinality issues caused by typos or inconsistent naming.
    * In this stage we removed two features and added five more.
2. Exploratory Data Analysis:
    * We conducted univariate analysis to understand distributions and multivariate analysis to investigate relationships between features and the target variable, guiding our subsequent decisions.
3. Feature Selection Strategy:
    * To identify the most relevant predictors, we performed an initial pass of imputation and scaling.
    * We then applied multiple feature selection techniques (Filter, Wrapper, and Embedded methods). Based on these insights, less informative features were removed to reduce noise and dimensionality.
4. Pipeline Construction:
    * We established a robust preprocessing pipeline to handle the selected features, applying RobustScaler for numerical features, KNNImputer for missing values, and TargetEncoder for categorical variables.
5. Model Development and Tuning:
    * Benchmarking: We initially trained and evaluated a diverse set of regression models.
    * Hyperparameter Tuning: We utilized RandomizedSearchCV with a PredefinedSplit to optimize the hyperparameters of the top-performing models, ensuring the tuning process strictly respected the train/validation split boundaries.
    * Ensemble Strategy: To further boost performance, we developed a Stacking Regressor. This final model combined the strengths of multiple powerful learners, such as Random Forest, Gradient Boosting, Extra Trees, Multilayer perceptron and HistGradientBoosting to achieve the best predictive results.

**Model Assessment Approach**

To rigorously evaluate our models and prevent overfitting, we adopted the Hold-out Method:
*   Data Splitting: The labeled dataset (train.csv) was split into a Training Set (75%) and a Validation Set (25%).
*   Validation Protocol: All intermediate evaluations, feature selection, and hyperparameter tuning decisions were made using the Training Set and validated against the Validation Set.
*   Final Evaluation: The effectiveness of the final Stacking model was confirmed by its lower MAE metric in the unseen test set (on kaggle) compared to the individual base models.

1. **log_mileage**
Mileage is highly right-skewed. Applying a logarithmic transformation reduces skewness, limits the impact of extreme values, and helps models better learn the relationship between mileage and price.

2. **car_age**
Vehicle depreciation depends on age rather than production year. Converting year into car age creates a more meaningful variable that directly captures depreciation effects.

3. **age_squared**
Car depreciation is non-linear, with faster value loss in early years. The squared age term allows models to capture this curvature without requiring complex non-linear models.

4. **miles_per_year**
Total mileage does not reflect usage intensity. Dividing mileage by car age distinguishes between lightly and heavily used vehicles. Using log-transformed mileage further reduces sensitivity to outliers.

5. **power_efficiency**
This feature captures the balance between engine power and fuel efficiency, which influences both performance perception and running costs, and therefore impacts resale price.

```python
plt.figure(figsize=(15, 5))
for i, col in enumerate(metric_features):
    plt.subplot(4, 3, i + 1)  
    sns.boxplot(x=train[col])
    plt.title(col)
    plt.xlabel("")

plt.tight_layout()
plt.show()
```

We observe a substantial number of points falling outside the interquartile range.
The prevalence of these outliers provides a strong justification for selecting the **RobustScaler** in our final pipeline. Unlike the **StandardScaler** (which is sensitive to the mean and variance distortions caused by outliers) or **MinMaxScaler** (which compresses the majority of data into a tiny range), the **RobustScaler** scales data based on percentiles. This ensures our model remains resilient to extreme values without necessitating excessive data removal.

Variables like **year** and **car_age** show mirrored outliers, as expected. The **age_squared** feature intentionally amplifies these distances to help the model capture non-linear depreciation patterns.

```python
sp_rows = 3
sp_cols = 4

# Prepare figure and axes
fig, axes = plt.subplots(sp_rows, sp_cols, figsize=(20, 8), tight_layout=True)

# Plot data with seaborn
features = [f for f in metric_features]

for ax, feat in zip(axes.flatten(), features):
    sns.histplot(train[feat], bins=30, kde=True, ax=ax)
    ax.set_ylabel('')


title = "Metric Variables"

fig.suptitle(title, fontsize=16, y=1.02)

plt.show()
```

The histograms reveal critical insights into the data structure that guided our preprocessing choices:
* Right-Skewness: The majority of the original variables, such as mileage, mpg, car_age, and miles_per_year, exhibit a strong right-skew. This indicates that most data points are concentrated at lower values, with a long tail of higher values.

* The impact of the logarithmic transformation is clearly visible. While the original mileage variable is heavily skewed, the engineered log_mileage feature follows a nearly Normal distribution. This transformation successfully stabilized the variance, making the feature more suitable for linear-based computations.

* The tax variable displays a highly irregular and multimodal distribution with distinct spikes rather than a continuous flow. This observation supports our subsequent decision to remove tax during the Feature Selection phase due to its noise and potentially low predictive power for a regression task.

#### Analysis

To reduce dimensionality to prevent overfitting, minimize noise, and improve the generalization capability of the final Stacking model, we decided to follow with feature selection.

To select the optimal feature set, we analyze feature importance through three different mathematical approaches, to get a more generalized feature selection:

1. **Multicollinearity check (Spearman correlation):** the goal is to identify redundant features that add complexity without adding new information.

2. **Linear Importance (Lasso + RFE):** the goal is to detect features with strong linear relationships with the price. We will apply these methods only to numeric features to avoid the biases of encoding categorical variables in linear models.

3. **Non-Linear Consensus (Tree-based ensemble):** the goal is to have a final veredict, as these models capture complex, non-linear patterns and interactions that linear models miss. We will aggregate importance from Random Forest, Extra Trees, and Gradient Boosting. All three share the same mechanism for calculating importance, making their results directly comparable in a single chart. Although HistGradientBoosting and KNN are part of our final stacking regressor, they were excluded from the feature selection visualization because the first lacks the ".feature_importances_" attribute and the second because it operates on geometric distances, not decision splits (comparing distance-based importance with tree-based importance would be inconsistent).

We acknowledge a potential inductive bias in utilizing tree-based ensembles for both feature selection and final prediction, which naturally optimizes the dataset for the specific mathematical assumptions of decision trees. To mitigate this and aim for a general solution, we avoided relying exclusively on tree importance and used this multi-view approach.

```python
X_num_FS = X_train_num_scaled
X_cat_FS = X_train_scaled[non_metric_features] 
num_cols = X_num_FS.columns.tolist()

# spearman correlation matrix
cor_spearman = X_num_FS.corr(method='spearman')

#  calculates feature importances
df_importance = calculate_importances(X_num_FS, X_cat_FS, y_train)

# 1. spearman + 2. lasso + rfe (linear) and 3. randomforest + extratrees + gradboosting (both have feature_importance_)
plot_dashboard_strategy(df_importance, cor_spearman, num_cols)
```

While the normalized plots are useful for comparing rankings across different models, they obscure the true magnitude of contribution. To make more informed removal decisions, we also analyzed the real feature importance values (which sum to 1), representing the actual percentage of predictive power.

```python
X_num_cat = pd.concat([X_num_FS, X_cat_FS], axis=1)
text_cols = X_num_cat.select_dtypes(include=['object', 'category']).columns
if len(text_cols) > 0:
    encoder = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)
    X_num_cat[text_cols] = encoder.fit_transform(X_num_cat[text_cols].astype(str))

df_final = df_importance.copy()
df_final['RandomForest'] = _get_model_importance(RandomForestRegressor, X_num_cat, y_train)
df_final['ExtraTrees'] = _get_model_importance(ExtraTreesRegressor, X_num_cat, y_train)
df_final['GradBoosting'] = _get_model_importance(GradientBoostingRegressor, X_num_cat, y_train)

print("\nNon-linear real feature importances:")
cols_trees = ['RandomForest', 'ExtraTrees', 'GradBoosting']

df_final['Mean'] = df_final[cols_trees].mean(axis=1) # sorts by the average mean
display(df_final[cols_trees + ['Mean']].sort_values(by='Mean', ascending=False).drop(columns=['Mean']))
```

## III. Regression Benchmarking

### III.I Pipeline

To ensure a fair and rigorous comparison across all candidate regression models, we used a standardized and optimized preprocessing pipeline. The flowchart of our pipeline is available on the 'README.md' file.
* KNNImputer was used to replace missing numerical data points by the mean value of their 10 nearest neighbors; 
* RobustScaler was applied to minimize the effect of outliers; 
* TargetEncoder was used to replace each category with the mean of the target (price). 

The performance impact of using KNN instead of a simple imputer and using TargetEncoder instead of an OrdinalEncoder is analyzed in the open ended section V.I.

```python
numeric_transformer = Pipeline(steps=[
    ('imputer', KNNImputer(n_neighbors=10)), # replace missing numerical data points by the mean value of their 10 nearest neighbors
    ('scaler', RobustScaler()) # normalizes the features mitigating the effect of outliers
])
categorical_transformer = Pipeline(steps=[
    ('target_encoder', TargetEncoder( # replaces categorical with the mean of the price for that category
        target_type="continuous", # regression problem
        smooth="auto", # weighted average between the category mean and the global mean price; increases generalization capability
        random_state=42
    ))
])
```

This preprocessor ensures that imputation and scaling are applied only to numerical data, while the encoding is applied only to categorical data:

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, metric_features),
        ('cat', categorical_transformer, non_metric_features)
    ],
)
```

A critical rule in this pipeline is to prevent data leakage. The .fit() method, which is used to learn the parameters (like the median or the scaling range), must be called only on the train data.

The same fitted object is then used to .transform() all three datasets (X_train, X_val, and X_test). This ensures that no information from the validation or test sets "leaks" into the training process, which would otherwise lead to an overfitted model and an artificially optimistic performance score.

```python
preprocessor.fit(X_train, y_train)

#transform all X
X_train_FINAL = preprocessor.transform(X_train)
X_val_FINAL = preprocessor.transform(X_val)
X_test_FINAL = preprocessor.transform(X_test)
```

Next, we concatenate the training and validation set and define a predifined split. We mark the train observations with a -1 flag and the val observations with 0. The main goal with this is to ensure that the subsequent benchmarking steps will always train on the original training set and evaluate on the original validation set, preventing data leakage.

```python
# Combine train and validation datasets
X_combined = np.concatenate([X_train_FINAL, X_val_FINAL])
y_combined = np.concatenate([y_train, y_val])

# Create a test fold index (-1 for train, 0 for validation)
test_fold = [-1] * len(X_train_FINAL) + [0] * len(X_val_FINAL)

print('Test fold: ', len(test_fold))
print('X_combined: ', len(X_combined))
print('y_combined: ', len(y_combined))

# Define the PredefinedSplit
ps = PredefinedSplit(test_fold=test_fold)
```

### III.IV Stacking

Despite the good performance of the individual Gradient Boosting Regressor, the final model chosen for submission is a stacked ensemble, aiming to reduce the MAE.

The final configuration of the stacking was determined through trial and error, monitored by iterative submissions on Kaggle. This optimization process ensured the stacking was as robust and effective as possible. The base models were adjusted through practical testing to test the impact of including or excluding certain models, such as the MLPRegressor, on overall ensemble performance. The meta-learner, responsible for blending the base predictions, was tested with various regularization algorithms, including RidgeCV, LassoCV, and ElasticNetCV, until the RidgeCV provided the optimal blending result.

```python
rf_best = models["Random Forest"]
gb_best = models["Gradient Boosting"]
et_best = models["Extra Trees"]
hgb_best = models["Hist Gradient Boosting"]
mlp_best = models["MLPRegressor"]

cv_meta = KFold(n_splits=5, shuffle=True, random_state=42)

final_layer = Pipeline([
    ("scaler", StandardScaler()),
    ("ridge", RidgeCV(
        alphas=[0.01, 0.1, 1.0, 5.0, 10.0],
        cv=cv_meta
    ))
])

stacking_model = StackingRegressor(
    estimators=[
        ('gb', gb_best),
        ('et', et_best),
        ('hgb',hgb_best),
        ('rf',rf_best),
        ('mlp',mlp_best)
    ],
    final_estimator=final_layer,
    cv=cv_meta,
    passthrough= True,
    n_jobs=1,
)

final_model_log = TransformedTargetRegressor(

    regressor=stacking_model,

    func=np.log1p,  # price to log

    inverse_func=np.expm1  # normal price

)
print("Training Final Stacking...")
final_model_log.fit(X_combined, y_combined)
print("Ready to Kaggleit!")
```

The final_model_log pipeline incorporates the main ensemble (StackingRegressor) that utilizes the five best optimized non-linear models (Gradient Boosting, Extra Trees, Random Forest, Hist Gradient Boosting, and MLPRegressor) as its base estimators and the final estimator (RidgeCV) uses regularization to optimally weight the contribution of each base prediction.

The entire pipeline was wrapped in a logarithmic transformation to normalize the distribution of vehicle prices and stabilize the variance. The inverse function ensures the final output is correctly reverted back to the true monetary scale.

### IV.I Ablation Study

In this section, the objective is to measure the individual contribution of each component of our data engineering pipeline, so we tested if techniques like KNNImputer and TargetEncoder justify their complexity compared to simpler alternatives. We compared our final pipeline against two simplified variants using a stratified sample of the training data.

```python
# using n_smaples=15000
X_sample, y_sample = resample(X_train, y_train, n_samples=15000, random_state=42)

# using only 1 model (random forest) to test the isolated effect of pre-processing
tester_model = RandomForestRegressor(n_estimators=50, max_depth=10, random_state=42, n_jobs=-1)

# our current pipeline
# KNN Imputer + Robust Scaler + Target Encoder
pipeline_current = ColumnTransformer(
    transformers=[
        ('num', Pipeline([('imputer', KNNImputer(n_neighbors=5)), ('scaler', RobustScaler())]), metric_features),
        ('cat', TargetEncoder(target_type="continuous"), non_metric_features)
    ]
)

# Simple Imputer + Robust Scaler + Target Encoder (what is the contribution of KNN?)
pipeline_noKNN = ColumnTransformer(
    transformers=[
        ('num', Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', RobustScaler())]), metric_features),
        ('cat', TargetEncoder(target_type="continuous"), non_metric_features)
    ]
)

# KNN Imputer + Robust Scaler + Ordinal Encoder (what is the contribution of Target Encoder?)
pipeline_noTarget = ColumnTransformer(
    transformers=[
        ('num', Pipeline([('imputer', KNNImputer(n_neighbors=5)), ('scaler', RobustScaler())]), metric_features),
        ('cat', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1), non_metric_features)
    ]
)

scenarios = {
    "1. Baseline (KNN + TargetEnc)": pipeline_current,
    "2. Simple Imputer (No KNN)": pipeline_noKNN,
    "3. Ordinal Encoder (No TargetEnc)": pipeline_noTarget
}

results = []

for name, preprocessor_openended in scenarios.items():
    full_pipeline = Pipeline([
        ('preprocessor', preprocessor_openended),
        ('model', tester_model)
    ])
    
    scores = cross_val_score(full_pipeline, X_sample, y_sample, cv=3, scoring='r2')
    
    for score in scores:
        results.append({'Pipeline': name, 'R2 Score': score})

plt.figure(figsize=(10, 5))
sns.boxplot(data=pd.DataFrame(results), x='Pipeline', y='R2 Score')
plt.title("What is the element of the pipeline that contributes more?")
plt.ylabel("R² Score")
plt.show()
```

The target encoder proved to be the most vital component of the pipeline. Replacing it with a standard ordinal encoder (3.) resulted in a significant performance drop on $R^2$. This confirms that high-cardinality features (like model and brand) contain strong price signals that are well captured by target averaging. The contribution of the KNN imputer was positive but marginal. When compared to a simple imputer (2.), the KNN showed a marginal improvement. Given our objective to maximize the model's performance, we decided to retain the original pipeline to maximize predictive accuracy, however small the margin.

We suspected that the model might struggle more with high-end vehicles simply because we have fewer examples of them in our training data compared to common cars. To test this, we calculated the average error (MAE) for three different market segments: economic cars (under £15.000), mid-range (£15.000-30.000), and luxury cars (over £30.000).

```python
# to optimize running time, we changed some parameters:
# n_samples=15000
X_sample, y_sample = resample(X_train_FINAL, y_train, n_samples=15000, random_state=42)

fast_stacking.fit(X_sample, y_sample)

y_pred_analysis = fast_stacking.predict(X_val_FINAL)

df_analysis = pd.DataFrame({'Real': y_val, 'Prediction': y_pred_analysis})
df_analysis['MAE'] = np.abs(df_analysis['Real'] - df_analysis['Prediction'])
df_analysis['MAPE'] = (np.abs(df_analysis['Real'] - df_analysis['Prediction']) / df_analysis['Real']) * 100

# price segments
bins = [0, 15000, 30000, 1000000]
labels = ['Economic (<15k)', 'Mid-range (15k-30k)', 'Luxury (>30k)']
df_analysis['Segment'] = pd.cut(df_analysis['Real'], bins=bins, labels=labels)

fig, axes = plt.subplots(1, 2, figsize=(16, 6)) 

# MAE (£))
sns.barplot(data=df_analysis, x='Segment', y='MAE', ax=axes[0], palette="Blues_d", errorbar=None)
axes[0].set_title("Absolute Error (MAE in £)")
axes[0].set_ylabel("Error in Euros")

# MAPE (%)
sns.barplot(data=df_analysis, x='Segment', y='MAPE', ax=axes[1], palette="Greens_d", errorbar=None)
axes[1].set_title("Relative Error (MAPE in %)")
axes[1].set_ylabel("Error Percentage (%)")

plt.tight_layout()
plt.show()

print("Performance by Segment:")
summary = df_analysis.groupby('Segment')[['MAE', 'MAPE']].mean()
print(summary.apply(lambda x: x.map(lambda v: f"{v:.2f}")))
```

Our model performs best in the mid-range segment (£15k-30k), achieving the lowest relative error (8.93% MAPE).
In the other hand, the economic segment (<£15k) shows the highest relative error (11.58% MAPE), despite having the lowest absolute error. This suggests that price variance in cheaper cars might be driven by unobserved factors (physical conditions like scratches or interior wear, for example) rather than technical factors. Finally, the luxury segment (>£30k) maintains a stable relative error (10.13%), showing that high-end pricing follows a more predictable, mathematical structure despite the larger absolute values involved.

### IV.IV General vs specialist model

In this section, we wanted to determine whether training a brand-specific model (specialist) yields better predictive accuracy than a single unified model (generalist). We selected the brand with the highest representation in the dataset.

```python
top_brand = X_train['Brand'].value_counts().idxmax()
print(f"Brand: {top_brand}")

mask_train = (X_train['Brand'] == top_brand).values
X_spec_train = X_train_FINAL[mask_train]
y_spec_train = y_train[mask_train]

# to ensure a fair comparison, we controlled the sample volume
X_general_train, y_general_train = resample(X_train_FINAL, y_train, n_samples=len(X_spec_train), random_state=42)

print(f"Generalist Model: {len(X_general_train)} cars (including other brands)")
print(f"Specialist Model: {len(X_spec_train)} cars ({top_brand})")

mask_val = (X_val['Brand'] == top_brand).values
X_val_spec = X_val_FINAL[mask_val]
y_val_spec = y_val[mask_val]

general_model = clone(fast_stacking)
general_model.fit(X_general_train, y_general_train)

specialist_model = clone(fast_stacking) 
specialist_model.fit(X_spec_train, y_spec_train)

y_pred_general = general_model.predict(X_val_spec)
mae_general = mean_absolute_error(y_val_spec, y_pred_general)

y_pred_spec = specialist_model.predict(X_val_spec)
mae_spec = mean_absolute_error(y_val_spec, y_pred_spec)

print(f"Final results for {top_brand}:")
print(f"Generalist error:  £{mae_general:.2f}")
print(f"Specialist error: £{mae_spec:.2f}")

diff = mae_general - mae_spec

if diff > 0:
    print(f"The best performance is achieved using a specific model. (Difference in MAE: £{diff:.2f})")
else:
    print(f"The best performance is achieved using a generalist model. (Difference in MAE: £{abs(diff):2f})")

results_df = pd.DataFrame({
    'Model Type': ['General Model', f'Specialist ({top_brand})'],
    'MAE (£)': [mae_general, mae_spec]
})

plt.figure(figsize=(6, 4))
sns.barplot(data=results_df, x='Model Type', y='MAE (£)', palette='viridis')
plt.title(f'General vs. Specialist ({top_brand})')
plt.ylabel('Mean Absolute Error')
plt.ylim(800, 1200)
plt.show()
```

The specialist model, trained exclusively on Ford vehicles, achieved a significantly lower MAE of £990.53 when predicting Ford prices, compared to the generalist model's MAE of £1208.02. This difference of £217.48 proves that the specialist model was better able to capture brand-specific nuances in depreciation, model value, and features, leading to more accurate valuations for the dominant brand in the dataset, as we could expect.
